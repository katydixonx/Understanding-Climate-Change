# Data Audit and EDA: Bastin et al. (2019), City Analogues
**Paper:** Bastin J-F et al. (2019) Understanding climate change from a global analysis of city analogues.

**Data:** S2 & S3 Files from the correction, https://doi.org/10.1371/journal.pone.0224120

**Approach:** The full dataset is small (534 rows), so is audited in full with no sampling.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
```

```python
base = "https://raw.githubusercontent.com/katydixonx/Understanding-Climate-Change/main/data/"
future = pd.read_csv(base + "pone.0224120.s002.csv")
current = pd.read_csv(base + "pone.0224120.s003.csv")
```

```python
rio_now = current.loc[current["NAMEASCII"] == "Rio de Janeiro", "Annual_Mean_Temperature"]
rio_2050 = future.loc[future["NAMEASCII"] == "Rio de Janeiro", "Annual_Mean_Temperature"]
print("Rio, file S003:", rio_now.values)
print("Rio, file S002:", rio_2050.values)
```

**Finding:** The two supporting files are not labelled. Therefore comparing Rio de Janeiro's annual mean temperature: S003=23.6 and S002=24.5 shows that S002 is the warmer 2050 projection and S003 is present day climate.

```python
print("Current:", current.shape)
print("Future:", future.shape)
print(current.dtypes.value_counts())
current.head()
```

**Finding:** Both files have 534 rows x 57 columns. The paper reports 520 cities, there are 14 more rows than stated.

```python
bioclim = [
    "Annual_Mean_Temperature", "Mean_Diurnal_Range", "Isothermality",
    "Temperature_Seasonality", "Max_Temperature_of_Warmest_Month",
    "Min_Temperature_of_Coldest_Month", "Temperature_Annual_Range",
    "Mean_Temperature_of_Wettest_Quarter", "Mean_Temperature_of_Driest_Quarter",
    "Mean_Temperature_of_Warmest_Quarter", "Mean_Temperature_of_Coldest_Quarter",
    "Annual_Precipitation", "Precipitation_of_Wettest_Month",
    "Precipitation_of_Driest_Month", "Precipitation_Seasonality",
    "Precipitation_of_Wettest_Quarter", "Precipitation_of_Driest_Quarter",
    "Precipitation_of_Warmest_Quarter", "Precipitation_of_Coldest_Quarter"
]
print(len(bioclim), "variables; all present:", all(c in current.columns for c in bioclim))

other_cols = [c for c in current.columns if c not in bioclim]
print(other_cols)
```

**Finding:** All 19 WorldClim bioclimatic variables are present. The remaining 38 columns are mostly GIS/shapefile leftovers (e.g. `FID_`, `Shape_Area`, `G200_*`), population figures and a `.geo` column storing coordinates as JSON text. None of these are used in the analysis, which makes the files harder to interpret.

```python
missing = pd.DataFrame({
    "current": current.isna().sum(),
    "future": future.isna().sum()
})
missing[(missing["current"] > 0) | (missing["future"] > 0)]
```

**Finding:** Climate variables are missing for 11 cities (present) and 9 (2050). `CONTINENT` is blank for 16 cities and `REALM` for 420, so these fields are unreliable for grouping.

```python
merged = current.merge(
    future[["system:index"] + bioclim],
    on="system:index",
    suffixes=("_now", "_2050")
)

now_cols = [c + "_now" for c in bioclim]
fut_cols = [c + "_2050" for c in bioclim]

complete = merged.dropna(subset=now_cols + fut_cols).copy()
print("Cities with complete climate data in both periods:", len(complete))

dropped = merged.loc[~merged.index.isin(complete.index), "NAMEASCII"]
print(dropped.tolist())
```

**Finding:** No special missing-value codes (e.g. -999) or physically impossible values were found, and units are plain °C and mm.
Quirks found:
- Duplicate city names: Hyderabad (Pakistan/India) and San Jose (USA/Costa Rica). These are distinct cities, but the authors' R script uses names as row labels, which could cause mismatches.
- "Australia" and "Oceania" appear as separate continents.
- "Water" (Managua) and "Polar" ecological zones are not handled by the authors' zone grouping.

```python
print(current[bioclim].describe().T[["min", "max"]])

print("Values <= -999:", (current.select_dtypes("number") <= -999).sum().sum())

print(current.loc[current["NAMEASCII"].duplicated(keep=False),
                  ["NAMEASCII", "CONTINENT", "LATITUDE", "LONGITUDE"]])

print(current["CONTINENT"].value_counts(dropna=False))
print(current["gez_name"].value_counts(dropna=False))
```

```python
complete["temp_change"] = (complete["Annual_Mean_Temperature_2050"]
                           - complete["Annual_Mean_Temperature_now"])
print(complete["temp_change"].describe())
complete.loc[complete["temp_change"] < 0, ["NAMEASCII", "temp_change"]]
```

**Finding:** Mean projected warming is +2.0 °C (range −1.1 to +4.1 °C). Two cities, Sana'a and Maiduguri, show cooling.

```python
fig, ax = plt.subplots(figsize=(8, 5))
sns.kdeplot(complete["Annual_Mean_Temperature_now"], fill=True,
            label="Present (1970–2000)", ax=ax)
sns.kdeplot(complete["Annual_Mean_Temperature_2050"], fill=True,
            label="2050 (RCP 4.5)", ax=ax)
ax.set_xlabel("Annual mean temperature (°C)")
ax.set_ylabel("Density")
ax.set_title("Annual mean temperature across 518 cities: present vs 2050")
ax.legend()
plt.tight_layout()
plt.savefig("fig1_temp_distribution.png", dpi=150)
plt.show()
```

Figure 1: present vs 2050 annual mean temperature distribution (n=518); mean shift +2.0°C

```python
def simple_zone(z):
    if pd.isna(z):
        return "Other/unknown"
    for k in ["Tropical", "Subtropical", "Temperate", "Boreal"]:
        if z.startswith(k):
            return k
    return "Other/unknown"

complete["zone"] = complete["gez_name"].apply(simple_zone)

fig, ax = plt.subplots(figsize=(8, 5))
sns.scatterplot(data=complete, x="LATITUDE", y="temp_change",
                hue="zone", alpha=0.7, ax=ax)
ax.axhline(0, color="grey", linestyle="--")
ax.set_xlabel("Latitude (°)")
ax.set_ylabel("Warming, 2050 − present (°C)")
ax.set_title("Projected warming by latitude and ecological zone")
plt.tight_layout()
plt.savefig("fig2_warming_latitude.png", dpi=150)
plt.show()
```

Figure 2: warming vs latitude by ecological zone (n=518). Warming is greatest at northern mid-latitudes, and the sample is skewed towards the Northern Hemisphere. Sana'a and Maiduguri show cooling. This could be because the present and future data come from different WorldClim versions.

```python
corr = complete[now_cols].corr()
corr.index = corr.columns = bioclim

fig, ax = plt.subplots(figsize=(11, 9))
sns.heatmap(corr, cmap="coolwarm", vmin=-1, vmax=1, center=0, square=True, ax=ax)
ax.set_title("Correlation between the 19 present-day bioclimatic variables")
plt.tight_layout()
plt.savefig("fig3_correlation.png", dpi=150)
plt.show()

upper = corr.abs().where(np.triu(np.ones(corr.shape), k=1).astype(bool))
print("Variable pairs with |r| > 0.8:", (upper > 0.8).sum().sum())
```

Figure 3: Pearson correlation between the 19 present-day bioclimatic variables (n = 518). Temperature and precipitation variables form two clusters, and 21 pairs have |r| > 0.8 (e.g. precipitation of driest month vs driest quarter). This redundancy supports the authors' use of PCA, but keeping only four components may lose weaker signals such as precipitation seasonality.

## Audit summary: paper vs data
- **Number of cities: does not match.** Paper reports 520; the data has 534 raw rows and 518 with complete records.
- **Bioclimatic variables: matches.** All 19 present.
- **Missing-value codes: partially documented.** No special codes (e.g. -999) found, but 16 cities have blank climate data, which the paper doesn't mention.
- **Data availability: does not match.** The original statement said all data was in the paper; a later correction added the files, which are unlabelled.
